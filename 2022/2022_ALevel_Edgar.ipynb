```python
#task 1.2 # Bubble sort
def bubble(array):
    for i in range(len(array)):
        for i in range(len(array)-i-1):
            if array[i]>array[i+1]:
                array[i],array[i+1]=array[i+1],array[i]
    return array
result_list=task2_1()
bubble(result_list)

        
```

```python
def mergesort(array):
    if len(array) <= 1:
        return array
    mid = len(array) // 2
    left = array[:mid]
    right = array[mid:]
    return merge(mergesort(left), mergesort(right))
def merge(left, right):
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1 
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged
# main
def main():
    array = task2_1()
    while True:
        sorting = input("What sorting algorithm do you want to use: B/M ").upper()
        if sorting == "B":
            print(bubble(array))
            return True
        elif sorting == "M":
            print(mergesort(array))
            return True
main()
    
```

```python
#task 2.4
def binary_search_recursive(array,target,low,high):  #only works on sorted list 
    if high>=low: #base case so hiugh doesent exceed low
         
        mid=(low+high)//2
        if array[mid]==target:
            return mid
        elif array[mid]<target:
            return binary_search_recursive(array,target,low=mid+1,high=high)
        elif array[mid]>target:
            return binary_search_recursive(array,target,low,high=mid-1)
    else:
        return -1
array=task2_1()
#print(binary_search_recursive(bubble(array),30,0,len(array)-1))

#actual case
integer=int(input("Enter an integer"))
print(bubble(array))
result=binary_search_recursive(bubble(array),integer,0,len(array)-1)
if result==-1:
    print("Its not in the list")
else:
    print(result)
```
