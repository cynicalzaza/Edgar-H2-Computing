**Task 2.1**

Write a function `task2_1(filename, quantity, maximum)` that:

- accepts three parameters:
  - `filename`, a string representing the name of a file
  - `quantity`, an integer representing the number of random integers to generate
  - `maximum`, representing the largest value that a random integer can take
- generates `quantity` random numbers between 0 and `maximum` (inclusive)
- writes those values, one per line, to a file named `filename`. **[4]**

Generate 1000 random numbers between 0 and 5000 (inclusive) and save them to a file called
`randomnumbers_<your name>_<centre number>_<index number>.txt` **[2]**


````python
# Task 2.1
def task2_1(filename,quantity,maxinum):
    import random
    with open(filename,"w") as file: #automatically closes the file
        for i in range (quantity):
            x=random.randint(0,maxinum)
            file.write(str(x)+ "\n")
#main            
task2_1("randomnumbers_edgar_25S22",1000,5000)
````

**Task 2.2**

Write a function `task2_2(list_of_integers)` that:

- takes a list of integers, `list_of_integers`
- sorts them into ascending order using merge sort
- returns the sorted list. **[7]**

Use the list `[56,25,4,98,0,18,4,5,7,0]` to test your function.

For example, the condition

`task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98]`

should return `True`. **[2]**


````python
# Task 2.2
#task2.2
#classic merge sort ascending order no tricks
def task2_2(list_of_intgers):
    if len(list_of_intgers) <= 1:
        return list_of_intgers
    mid = len(list_of_intgers) // 2
    # Capture the sorted slices returned by the recursive calls
    left_half = task2_2(list_of_intgers[:mid])
    right_half = task2_2(list_of_intgers[mid:])
    i = j = k = 0
    while i < len(left_half) and j < len(right_half):
        if left_half[i] < right_half[j]:
            list_of_intgers[k] = left_half[i]
            i += 1
        else:
            list_of_intgers[k] = right_half[j]
            j += 1
        k += 1
    while i < len(left_half):
        list_of_intgers[k] = left_half[i]
        i += 1
        k += 1
    while j < len(right_half):
        list_of_intgers[k] = right_half[j]
        j += 1
        k += 1
    return list_of_intgers
task2_2([56,25,4,98,0,18,4,5,7,0])
````

````python
# Task 2.3
#task2.3
#read file,convert string to integer,result= task2_2 ,write result into newfile
def task2_3(filename_in,filename_out):
    with open(filename_in,"r") as file: #automatically closes the file
        contents=[]
        for line in file:
            newline=line.strip().split(",")
            contents.append(int(newline[0]))                                    
            result=task2_2(contents)
            with open (filename_out,"w") as file:
                file.write(str(result)) 
    file.close()
task2_3("randomnumbers_edgar_25S22","sortednumbers_edgar")
````
